```python
import nibabel as nib
import matplotlib.pyplot as plt

file_path = "/Users/rohithasresta/Desktop/sub-apat101/ses-01/anat/sub-apat101_ses-01_acq-2DrestHRref_defaced_T1w.nii"

img = nib.load(file_path)
data = img.get_fdata()

print("Shape:", data.shape)
```

```python
import matplotlib.pyplot as plt

slice_idx = data.shape[2] // 2

plt.imshow(data[:, :, slice_idx], cmap='gray')
plt.title("Middle MRI Slice")
plt.axis('off')
plt.show()
```

```python
fig, axes = plt.subplots(1, 5, figsize=(15,5))

for i, ax in enumerate(axes):
    idx = int(data.shape[2] * (i+1) / 6)
    ax.imshow(data[:, :, idx], cmap='gray')
    ax.axis('off')

plt.suptitle("MRI Slice Visualization")
plt.show()
```

```python
print("Data shape:", data.shape)
print("Number of slices:", data.shape[2])
print("Image resolution:", data.shape[0], "x", data.shape[1])
```

```python
print("Min intensity:", data.min())
print("Max intensity:", data.max())
print("Mean intensity:", data.mean())
```
