```python
import nibabel as nib
import matplotlib.pyplot as plt
import numpy as np

print("Libraries loaded!")
```

```python
DATA_ROOT = '/home/fabric/work/data'

scans = {
    'sub-pat02_ses-01': f'{DATA_ROOT}/sub-pat02/ses-01/anat/sub-pat02_ses-01_acq-3D_defaced_T1w.nii.gz',
    'sub-pat02_ses-05': f'{DATA_ROOT}/sub-pat02/ses-05/anat/sub-pat02_ses-05_acq-3D_defaced_T1w.nii.gz',
    'sub-pat03_ses-01': f'{DATA_ROOT}/sub-pat03/ses-01/anat/sub-pat03_ses-01_acq-3D_defaced_T1w.nii.gz',
    'sub-pat03_ses-07': f'{DATA_ROOT}/sub-pat03/ses-07/anat/sub-pat03_ses-07_acq-3D_defaced_T1w.nii.gz',
}

print("Checking files...")
for name, path in scans.items():
    import os
    status = '✓' if os.path.exists(path) else '✗ MISSING'
    print(f'  {status}  {name}')
```

```python
for name, path in scans.items():
    img = nib.load(path)
    data = img.get_fdata()
    print(f'--- {name} ---')
    print(f'  Shape:      {data.shape}')
    print(f'  Resolution: {img.header.get_zooms()} mm')
    print(f'  Mean intensity: {data[data > 0].mean():.1f}')
    print()
```

```python
fig, axes = plt.subplots(4, 5, figsize=(15, 12))
fig.suptitle('T1w Brain Slices — All Sessions', fontsize=14, fontweight='bold')

for row, (name, path) in enumerate(scans.items()):
    data = nib.load(path).get_fdata()
    for col in range(5):
        idx = int(data.shape[2] * (col + 1) / 6)
        axes[row, col].imshow(data[:, :, idx].T, cmap='gray', origin='lower')
        if col == 0:
            axes[row, col].set_ylabel(name, fontsize=8, rotation=45, ha='right')
        axes[row, col].set_title(f'z={idx}', fontsize=7)
        axes[row, col].axis('off')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

colors = ['blue', 'royalblue', 'red', 'salmon']
for (name, path), color in zip(scans.items(), colors):
    data = nib.load(path).get_fdata()
    flat = data[data > 0].flatten()
    ax.hist(flat, bins=100, alpha=0.5, label=name, color=color)

ax.set_xlabel('Voxel Intensity')
ax.set_ylabel('Count')
ax.set_title('T1w Intensity Distribution — QA Check')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

colors = ['blue', 'royalblue', 'red', 'salmon']
for (name, path), color in zip(scans.items(), colors):
    data = nib.load(path).get_fdata()
    flat = data[data > 0].flatten()
    # Normalize to 0-1 range
    flat_norm = (flat - flat.min()) / (flat.max() - flat.min())
    ax.hist(flat_norm, bins=100, alpha=0.5, label=name, color=color)

ax.set_xlabel('Normalized Voxel Intensity (0-1)')
ax.set_ylabel('Count')
ax.set_title('T1w Intensity Distribution — Normalized (Post-QA)')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

rows = []
for name, path in scans.items():
    img = nib.load(path)
    data = img.get_fdata()
    flat = data[data > 0].flatten()
    subject, session = name.split('_')
    rows.append({
        'Subject': subject,
        'Session': session,
        'Shape': str(img.shape),
        'Resolution (mm)': f"{img.header.get_zooms()[0]:.2f}",
        'Raw Mean Intensity': f"{flat.mean():.0f}",
        'Raw Std': f"{flat.std():.0f}",
        'Non-zero Voxels': len(flat),
    })

df = pd.DataFrame(rows)
print(df.to_string(index=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Brain measurements from DeepPrep output
data = {
    'Subject': ['sub-pat02', 'sub-pat03'],
    'Mean Cortical Thickness (mm)': [2.50081, 2.25745],
    'White Surface Area (mm²)': [79946.3, 69094.4],
    'Brain Segmentation Volume (mm³)': [1151730, 955818],
    'Total Intracranial Volume (mm³)': [1429789, 1037475],
}

df = pd.DataFrame(data)
print(df.to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 5))
fig.suptitle('DeepPrep Output: Brain Measurements\nsub-pat02 vs sub-pat03 (Right TLE)', 
             fontsize=13, fontweight='bold')

metrics = [
    'Mean Cortical Thickness (mm)',
    'White Surface Area (mm²)',
    'Brain Segmentation Volume (mm³)',
    'Total Intracranial Volume (mm³)',
]

colors = ['steelblue', 'salmon']

for ax, metric in zip(axes, metrics):
    bars = ax.bar(df['Subject'], df[metric], color=colors, width=0.5)
    ax.set_title(metric, fontsize=9, fontweight='bold')
    ax.set_ylabel(metric.split('(')[1].replace(')', '') if '(' in metric else '')
    ax.set_xticks(range(len(df)))
    ax.set_xticklabels(df['Subject'], rotation=15, ha='right', fontsize=8)
    # Add value labels on bars
    for bar, val in zip(bars, df[metric]):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() * 0.95,
                f'{val:,.1f}', ha='center', va='top', fontsize=8, color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# All three subjects - epilepsy vs healthy control
data = {
    'Subject': ['sub-pat02\n(Epilepsy)', 'sub-pat03\n(Epilepsy)', 'sub-con02\n(Healthy)'],
    'Mean Cortical Thickness (mm)': [2.50081, 2.25745, 2.20093],
    'White Surface Area (mm²)': [79946.3, 69094.4, 77792.2],
    'Brain Segmentation Volume (mm³)': [1151730, 955818, 1027026],
    'Total Intracranial Volume (mm³)': [1429789, 1037475, 1071308],
}

df = pd.DataFrame(data)

fig, axes = plt.subplots(1, 4, figsize=(18, 5))
fig.suptitle('DeepPrep Output: Epilepsy Patients vs Healthy Control\n(sub-pat02, sub-pat03 = Right TLE | sub-con02 = Healthy)', 
             fontsize=12, fontweight='bold')

colors = ['steelblue', 'steelblue', 'green']
metrics = [
    'Mean Cortical Thickness (mm)',
    'White Surface Area (mm²)',
    'Brain Segmentation Volume (mm³)',
    'Total Intracranial Volume (mm³)',
]

for ax, metric in zip(axes, metrics):
    bars = ax.bar(df['Subject'], df[metric], color=colors, width=0.5, alpha=0.8)
    ax.set_title(metric, fontsize=9, fontweight='bold')
    ax.set_xticks(range(len(df)))
    ax.set_xticklabels(df['Subject'], fontsize=8)
    for bar, val in zip(bars, df[metric]):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() * 0.95,
                f'{val:,.1f}', ha='center', va='top', fontsize=7, color='white', fontweight='bold')

plt.tight_layout()
plt.show()
print("Blue = Epilepsy | Green = Healthy Control")
```
